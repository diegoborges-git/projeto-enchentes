```python
import pandas as pd

# Caminho para o ficheiro extraído na pasta dados/
arquivo_txt = "../dados/CAGEDMOV202404.txt"

poa_chunks = []
print("A processar e filtrar o ficheiro do CAGED em blocos...")

chunk_size = 100000
for chunk in pd.read_csv(
    arquivo_txt,
    sep=";",
    encoding="utf-8-sig",
    chunksize=chunk_size,
    low_memory=False,
):
  # Converte a coluna para numérico para garantir a comparação exata
  chunk["município"] = pd.to_numeric(chunk["município"], errors="coerce")

  # No Novo CAGED, o código de Porto Alegre usa 6 dígitos: 431490
  poa_subset = chunk[chunk["município"] == 431490]

  if not poa_subset.empty:
    poa_chunks.append(poa_subset)

# Junta todos os pedaços filtrados num único DataFrame
if poa_chunks:
  df_poa = pd.concat(poa_chunks, ignore_index=True)
  print(
      f"Filtro concluído com sucesso! Encontrados {len(df_poa)} registos para"
      " Porto Alegre."
  )

  # Salva o subset otimizado para o projeto
  df_poa.to_csv("caged_poa_abril_2024.csv", index=False)
  print("Ficheiro 'caged_poa_abril_2024.csv' guardado com sucesso!")

  # Exibe uma prévia das primeiras linhas
  display(df_poa.head(3))
  #display(df_poa.info())
  #display(df_poa.describe())
  # Lista todas as colunas presentes no DataFrame do CAGED
  print(df_poa.columns.tolist())
else:
  print("Nenhum registo encontrado.")
```

```python
import pandas as pd

# Caminho para o ficheiro .COMT na pasta de originais
arquivo_comt = "../dados/originais/RAIS_ESTAB_PUB.COMT"

poa_estab_chunks = []
chunk_size = 100000

print(
    "A processar a base nacional da RAIS em blocos (para evitar travamentos)..."
)

# Lê o ficheiro em blocos de 100 mil linhas
for i, chunk in enumerate(
    pd.read_csv(
        arquivo_comt,
        sep=",",
        encoding="latin1",
        chunksize=chunk_size,
        low_memory=False,
    )
):
  # Garante que a coluna do município é tratada como string limpa
  chunk["Município - Código"] = (
      chunk["Município - Código"].astype(str).str.strip()
  )

  # Filtra direto no bloco: UF 43 (RS) e Município de Porto Alegre (431490...)
  sub_poa = chunk[
      (chunk["UF - Código"] == 43)
      & (chunk["Município - Código"].str.contains("431490"))
  ]

  if not sub_poa.empty:
    poa_estab_chunks.append(sub_poa)

# Junta todos os pedaços filtrados de Porto Alegre
if poa_estab_chunks:
  df_poa_estab = pd.concat(poa_estab_chunks, ignore_index=True)
  print(
      f"\nSucesso! Encontrados {len(df_poa_estab)} estabelecimentos em Porto"
      " Alegre."
  )

  # Guarda num CSV leve e exclusivo para o teu projeto
  df_poa_estab.to_csv("rais_poa_estabelecimentos.csv", index=False)
  print("Ficheiro 'rais_poa_estabelecimentos.csv' guardado com sucesso!")

  # Mostra uma prévia
  display(df_poa_estab.head(3))
else:
  print("\nNenhum estabelecimento encontrado para Porto Alegre.")
```

```python
import pandas as pd

# 1. Carrega os estabelecimentos de Porto Alegre
df_poa = pd.read_csv("rais_poa_estabelecimentos.csv", low_memory=False)

# Trata o CEP para garantir 8 dígitos
df_poa["CEP_str"] = (
    df_poa["CEP Estab"].fillna(0).astype(int).astype(str).str.zfill(8)
)
df_poa["CEP5"] = df_poa["CEP_str"].str[:5].astype(int)


# 2. Dicionário refinado de faixas de CEP para os principais bairros de Porto Alegre
def mapear_todos_bairros(c):
  # Centro e Adjacências
  if 90000 <= c <= 90099:
    return "Centro Histórico"
  elif 90100 <= c <= 90130:
    return "Menino Deus"
  elif 90150 <= c <= 90160:
    return "Praia de Belas"
  elif 90400 <= c <= 90430:
    return "Moinhos de Vento"
  elif 90440 <= c <= 90470:
    return "Rio Branco"
  elif 90480 <= c <= 90520:
    return "Auxiliadora"
  elif 90530 <= c <= 90570:
    return "Bela Vista / Higienópolis"
  elif 90600 <= c <= 90699:
    return "Santana / Jardim Botânico"

  # Quarto Distrito e Zona Norte / Eixo Industrial afetado pelas cheias
  elif 90200 <= c <= 90210:
    return "Navegantes"
  elif 90220 <= c <= 90240:
    return "São Geraldo"
  elif 90250 <= c <= 90280:
    return "Floresta"
  elif 90300 <= c <= 90340:
    return "São João"
  elif 90350 <= c <= 90360:
    return "Passo d'Areia"
  elif 91000 <= c <= 91090:
    return "Sarandi / São Sebastião"
  elif 91100 <= c <= 91180:
    return "Anchieta / Humaitá / Farrapos"

  # Outras zonas importantes de POA
  elif 90800 <= c <= 90840:
    return "Medianeira / Cristal"
  elif 90850 <= c <= 90899:
    return "Teresópolis / Camaquã"
  elif 91300 <= c <= 91370:
    return "Jardim Itú / Passo das Pedras"
  elif 91400 <= c <= 91450:
    return "Bom Jesus / Protásio Alves"
  elif 91500 <= c <= 91540:
    return "Agronomia"
  elif 91700 <= c <= 91790:
    return "Cavalhada / Tristeza / Ipanema"
  elif 91900 <= c <= 91990:
    return "Belém Novo / Restinga"

  else:
    return "Demais Bairros / Outras Zonas de POA"


# Aplica a função detalhada
df_poa["Bairro_Detalhado"] = df_poa["CEP5"].apply(mapear_todos_bairros)

# 3. Agrupa e consolida para todos os bairros mapeados
tabela_94_bairros = (
    df_poa.groupby("Bairro_Detalhado")
    .agg(
        Total_Estabelecimentos=("CEP Estab", "count"),
        Total_Vinculos_Ativos=("Qtd Vínculos Ativos", "sum"),
    )
    .reset_index()
)

# Ordena por volume de vínculos
tabela_94_bairros = tabela_94_bairros.sort_values(
    by="Total_Vinculos_Ativos", ascending=False
)

print("=== DISTRIBUIÇÃO CONSOLIDADA DE EMPREGOS POR BAIRRO EM POA ===")
display(tabela_94_bairros)

# Guarda o ficheiro final detalhado
tabela_94_bairros.to_csv("empregos_detalhado_bairros_poa.csv", index=False)
print(
    "\nFicheiro 'empregos_detalhado_bairros_poa.csv' guardado com sucesso!"
)
```
